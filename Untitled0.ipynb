```python
import pandas as pd
```

```python
data={
    "CUSTOMER ID":[101,102,103,104,105],
    "name":["alice", "bob","step","hurt","sam"],
    "age":[25,30,35,40,45],
    "amount":[1200,1300,1400,237,124],
    "rating":[3.5,2,3,4,5]
    }
df=pd.DataFrame(data)
print(df)
```

```python
#first row
df.iloc[0]
```

```python
df.iloc[0:3]
```

```python
#first 5 rows and 3 column
df.iloc[0:5,0:3]
```

```python
df.loc[0:3,["CUSTOMER ID","amount","rating"]]
```

```python
df.loc[df["amount"]>500,["CUSTOMER ID","rating"]]
```
